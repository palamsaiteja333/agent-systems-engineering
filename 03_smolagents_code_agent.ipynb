```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, InferenceClientModel
```

```python
model = InferenceClientModel(
    model_id="Qwen/Qwen2.5-Coder-32B-Instruct"
)

print("Qwen/Qwen2.5-Coder-32B-Instruct model is configured")
```

```python
agent = CodeAgent(
    model=model,
    tools=[DuckDuckGoSearchTool()],
    max_steps=5,
    verbosity_level=2
)

print("agent is configured")
```

```python
task = "Search the web for the official Model Context Protocol documentation and explain in 3 bullet points what problem MCP is designed to solve for AI applications."
```

```python
response = agent.run(task)
print(response)
```

```python
from smolagents import tool


@tool
def recommend_incident_action(severity: str) -> str:
    """
    Recommend the response for a production incident.

    Args:
        severity: Incident severity. Supported values are
                  critical, high, medium, or low.
    """

    if severity == "critical":
        return "Page the on-call engineer and notify the incident commander."

    elif severity == "high":
        return "Assign the incident immediately and begin investigation."

    elif severity == "medium":
        return "Add the incident to the same-day investigation queue."

    elif severity == "low":
        return "Add the incident to the standard support queue."

    else:
        return "Unknown severity. Use critical, high, medium, or low."
```

```python
incident_agent = CodeAgent(
    tools= [recommend_incident_action],
    model=model,
    max_steps=5,
    verbosity_level=2
)
```

```python
response = incident_agent.run("critical")
print(response)
```

```python
task = """
Our production payment API is completely unavailable
and customers are unable to process transactions.

What incident response should we initiate?
"""

response = incident_agent.run(task)

print(response)
```

```python
incident_and_web_agent = CodeAgent(
    tools=[recommend_incident_action, DuckDuckGoSearchTool()],
    model=model,
    verbosity_level=2,
    max_steps=5
)
```

```python
task = """
Our production payment API is completely unavailable
and customers cannot process transactions.

Search for current incident-response best practices for a major
production service outage, then determine what response level
we should initiate using the incident-response tool available to you.

Keep the final answer concise.
"""
```

```python
response = incident_and_web_agent.run(task)
print(response)
```

```python
sla_agent = CodeAgent(
    tools=[],
    model=model,
    additional_authorized_imports=["datetime"],
    max_steps=5,
    verbosity_level=2
)

print("SLA agent created")
```

```python
task = """
A critical production incident was opened on
September 21, 2026 at 11:50 PM.

The response SLA is 30 minutes.

Use Python datetime operations to calculate the exact
response deadline, including the date and time.
"""

response = sla_agent.run(task)

print(response)
```

```python
task = """
Import the requests library and print its module name.

Do not use another approach. I specifically want you to
try importing requests.
"""

response = sla_agent.run(task)

print(response)
```

## What I observed

I intentionally asked the agent to import `requests` even though I had
only added `datetime` to `additional_authorized_imports`.

The model still generated the Python code:

`import requests`

but the execution environment rejected the action because `requests`
was not in the authorized import list.

What was interesting was that the agent did not stop after the failed
execution. The import error became part of the next step, and the agent
adjusted its response before eventually returning a final answer.

The execution flow looked like this:

Task
→ generated Python
→ unauthorized import blocked
→ error returned to the agent
→ agent handled the failure
→ final answer

My main takeaway is that the model being able to generate an action
does not mean the runtime has to allow that action.

For a code generating agent, I need to think about both:

- what the model is capable of generating
- what the execution environment is permitted to run

## What I learned

In this experiment, I went beyond giving an agent a single tool and
looked more closely at how `CodeAgent` generates and executes Python
actions.

The main things I understood were:

- an agent can choose between multiple available tools
- code actions can be used for calculations and multi-step tasks
- generated code and execution permissions are separate concerns
- an execution failure can become part of the next agent step instead
  of terminating the entire workflow

The unauthorized import experiment was the most useful part for me.
The model was able to generate `import requests`, but the runtime
rejected it because that import had not been authorized.

That made the separation between model capability and execution
permission much clearer to me.
